## Victor Djuma – R

In this lab, I learned how unsupervised learning is different from the supervised learning methods we used in earlier modules. In supervised learning, the model is trained with labeled data, which means we already know the correct target or answer. In unsupervised learning, there is no target variable. The algorithm looks for patterns, groups, or structure in the data by itself. This helped me understand that machine learning is not always about predicting a known answer. Sometimes the goal is to discover information that is hidden inside the data.

The first main concept I practiced was K-Means clustering. I learned that K-Means groups similar data points into clusters based on their distance from cluster centers called centroids. One challenge is deciding how many clusters, or k, should be used. To solve this, I used the Elbow Method. I tested values of k from 1 to 10 and plotted the inertia for each value. The graph showed an elbow around k = 4, which suggested that four clusters were a good choice for the synthetic customer data.

After identifying k = 4, I trained a K-Means model and visualized the results. The graph clearly showed four different groups, and the red X marks represented the centroid of each cluster. Seeing the clusters visually helped me understand the concept better than only reading about it. I could see how K-Means automatically separated the data into groups without being given labels.

The second major concept was Principal Component Analysis, or PCA. I learned that PCA is used for dimensionality reduction. It reduces the number of features while trying to preserve as much useful information as possible. In the lab, the Iris dataset originally had four features, and I reduced it to two principal components. The first two components explained about 97.77% of the total variance. This showed me that most of the important information from the original four features could still be represented using only two dimensions.

The PCA graph also helped me see how the Iris species were distributed after dimensionality reduction. I learned that PCA can make complex, high-dimensional data easier to visualize and can also reduce the amount of computation needed for machine learning tasks.

Overall, this lab helped me understand why unsupervised learning is useful when labels are not available. I learned how K-Means can discover natural groups in data and how PCA can simplify data while preserving most of its information. The most important lesson for me was understanding that clustering and dimensionality reduction solve different problems, but they can also work together to make complex datasets easier to analyze.
